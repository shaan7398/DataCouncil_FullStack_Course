```python
def even_odd(n):
    if n % 2 ==0:
        return(f"The number {n} is Even")
    else:
        return(f"The number {n} is Odd")
```

```python
even_odd(7)
```

```python
def max_of_three(n1,n2,n3):
    if ((n1 > n2) and (n1 > n3)):
        return(f"The number {n1} is highest of three")
    elif ((n2>n1) and (n2>n3)):
        return(f"The number {n2} is highest of three")
    else:
        return(f"The number {n3} is highest of three")
```

```python
max_of_three(1,27,97)
```

```python
def arithmatic_operation(a,b,operator):
    if(operator == '+'):
        return(a+b)
    elif(operator == '-'):
        return(a-b)
    elif(operator == '*'):
        return(a*b)
    elif(operator == '/'):
        return(a/b)
    else:
        return("Please give operator in quotes")
```

```python
arithmatic_operation(3,4,'+')
```

```python
def palindrome(a):
    palin_a = a[::-1]
    if(a == palin_a):
        return(f"The string is Palindrome {a}")
    else:
        return("Not a palindrome")
```

```python
palindrome("Ok")
```

```python
def number_of_vowels(a):
    """This is fucntion to count Vowels in a string
    Please provide the string as argument"""
    vowels = "a,e,i,o,u,A,E,I,O,U"
    count = 0

    for i in a:
        if( i in vowels):
            count = count+1
    return (f"The number of vowels in {a} are {count}")
    
```

```python
def Factorial(a):
    """This function is used to find the factorial
    please enter the number as argument"""
    factorial = 1
    for i in range(1,a+1):
        factorial = factorial * i
    return factorial
```

```python
Factorial(8)
```

```python
def sum_average(*args):
    sum = 0
    for i in args:
        sum = sum + i
    count = len(args)
    average = sum/count
    return(f"The sum of all numbers is {sum} annd Average is {average}")
```

```python
sum_average(1,2,3,4,5)
```
